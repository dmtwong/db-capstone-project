### b) Query from Client (interacting with Database)

Task 1: connect with your database, and Create a cursor


```python
load_dotenv()
connection = connector.connect(
    user=os.getenv("DB_USER"),
    password=os.getenv("DB_PASS"),
    database="littlelemonproject"
)
```

```python
cursor = connection.cursor()
```

Task 2: query the database to show all tables within the database, view the results of your query

```python
show_tables_query = "SHOW tables" 
cursor.execute(show_tables_query)
```

```python
results = cursor.fetchall()
print(results)
```

Task 3: require the full name and contact details for every customer that has placed an order greater than $60 for a promotional campaign. 

```python
join_tables_query = "SELECT b.cust_name, b.phone, b.email, a.Cost FROM Orders as A INNER JOIN CUST as b WHERE a.Cost > 60"
cursor.execute(join_tables_query)
```
